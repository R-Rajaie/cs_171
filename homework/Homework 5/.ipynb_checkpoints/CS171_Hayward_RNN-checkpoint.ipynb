
# Hayward Weather Forecast RNN Model

This notebook adapts the RNN workflow from the RNN lecture to predict the next seven days of **wind speed**, **precipitation**, and **average temperature** for the Bay Area using the Hayward airport weather dataset. The goal is to build a model that ingests 30 days of historical observations and outputs a seven‑day forecast for all three variables. The approach closely follows the structure of the provided lecture notebooks, but it is tailored to the `Hayward_Weather_Data.csv` data.

## Overview

1. **Load and clean the data** – read the CSV file, keep only the `AWND` (average wind speed), `PRCP` (precipitation), and `TAVG` (average temperature) columns, and remove rows with missing or bad values.
2. **Normalize the features** – compute the mean and standard deviation for each variable and scale them to zero mean and unit variance.
3. **Create sequences** – build input sequences of 30 days and target sequences of the following 7 days.
4. **Train an LSTM model** – use a PyTorch LSTM to model the temporal dynamics and predict multi‑step outputs.
5. **Evaluate and save the model** – monitor training/test losses and save both the model class (as a `.py` file) and the trained weights for use in scoring.

> **Note:** Running this notebook requires PyTorch. If you encounter an import error for `torch`, install it in your environment or run on a machine where PyTorch is available.


```

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
from torch import nn
from torch.utils.data import TensorDataset, DataLoader

torch.manual_seed(42)

# Configure device
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")
print("Using device:", device)
```


## 1. Load and clean the Hayward weather data

The dataset `Hayward_Weather_Data.csv` contains daily weather observations from the Hayward airport. We focus on three columns:

* **AWND** – average wind speed (knots)
* **PRCP** – precipitation (inches)
* **TAVG** – average temperature (°F)

Since this is a real‑world dataset, missing or bad values may appear. We remove any rows containing NaNs or obviously invalid measurements before proceeding.


```

# Path to the Hayward dataset
csv_path = 'Hayward_Weather_Data.csv'  # Ensure this file is in the working directory

# Load the CSV file
df = pd.read_csv(csv_path)

# Keep only the relevant columns
weather_df = df[['AWND', 'PRCP', 'TAVG']].copy()

# Drop rows with missing values
weather_df = weather_df.dropna().reset_index(drop=True)

# Display the first few rows
print(weather_df.head())
print(f"Total days after cleaning: {len(weather_df)}")
```


### Visualize the cleaned data

It’s good practice to visualize each variable to inspect seasonality, trends, and potential outliers. Here we plot the time series of wind speed, precipitation, and temperature for the first year in the record.


```

plt.figure(figsize=(10, 7))

plt.subplot(3,1,1)
plt.plot(weather_df['AWND'][:366], color='purple')
plt.ylabel('Wind Speed (kts)')

plt.subplot(3,1,2)
plt.plot(weather_df['PRCP'][:366], color='blue')
plt.ylabel('Precipitation (in)')

plt.subplot(3,1,3)
plt.plot(weather_df['TAVG'][:366], color='red')
plt.ylabel('Average Temp (°F)')
plt.xlabel('Days')

plt.tight_layout()
plt.show()
```


## 2. Normalize the features

We standardize each variable to zero mean and unit variance. Scaling improves numerical stability and accelerates gradient‑based training. The mean and standard deviation are stored for later use when de‑normalizing the predictions.


```

# Extract values as a NumPy array
features = weather_df[['AWND', 'PRCP', 'TAVG']].values

# Compute mean and std for each column
mean = features.mean(axis=0)
std = features.std(axis=0)

# Normalize
features_norm = (features - mean) / std

print('Means:', mean)
print('Standard deviations:', std)
```


## 3. Create input (30 days) and target (7 days) sequences

We slide a window across the normalized dataset. For each window of 30 consecutive days we record the following 7 days as the target. This results in arrays `X` and `y` of shapes `(num_samples, 30, 3)` and `(num_samples, 7, 3)` respectively.


```

# Window lengths
n_previous_days = 30
n_future_days = 7

# Construct sequences
X_seq, y_seq = [], []
for i in range(len(features_norm) - n_previous_days - n_future_days):
    X_seq.append(features_norm[i:i + n_previous_days])
    y_seq.append(features_norm[i + n_previous_days : i + n_previous_days + n_future_days])

X_seq = np.stack(X_seq)
y_seq = np.stack(y_seq)

print('Number of samples:', len(X_seq))
print('Input shape:', X_seq.shape)
print('Target shape:', y_seq.shape)
```


## 4. Split into training and test sets

We allocate the first 80% of the sequence samples for training and the remaining 20% for testing.


```

split_index = int(0.8 * len(X_seq))

X_train, y_train = X_seq[:split_index], y_seq[:split_index]
X_test, y_test   = X_seq[split_index:], y_seq[split_index:]

# Convert to PyTorch tensors
X_train_tensor = torch.tensor(X_train).float()
X_test_tensor  = torch.tensor(X_test).float()
y_train_tensor = torch.tensor(y_train).float()
y_test_tensor  = torch.tensor(y_test).float()

# Create data loaders
batch_size = 16
train_loader = DataLoader(TensorDataset(X_train_tensor, y_train_tensor), batch_size=batch_size, shuffle=True)
test_loader  = DataLoader(TensorDataset(X_test_tensor,  y_test_tensor),  batch_size=batch_size, shuffle=False)

print(f"Training samples: {len(X_train)}; Test samples: {len(X_test)}")
```


## 5. Build the LSTM model

We define a custom PyTorch module that uses an LSTM layer to process the 30‑day input. The last hidden state feeds into a fully connected layer that produces a `7 × 3` output for the future week.


```

class WeatherRNN(nn.Module):
    def __init__(self, n_future_days, input_size=3, hidden_size=64, num_layers=1):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )
        self.fc = nn.Linear(hidden_size, n_future_days * input_size)
        self.n_future_days = n_future_days
        self.input_size = input_size

    def forward(self, x):
        out, _ = self.lstm(x)
        last_hidden = out[:, -1, :]
        out_fc = self.fc(last_hidden)
        return out_fc.view(-1, self.n_future_days, self.input_size)

# Instantiate the model
n_future_days = 7
model = WeatherRNN(n_future_days=n_future_days)
model = model.to(device)

print(model)
```


## 6. Train the model

We use mean squared error (MSE) loss and the Adam optimizer. The training loop iterates over several epochs, computing loss on both the training and test sets. Adjust `EPOCHS` and `hidden_size` as needed.


```

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

EPOCHS = 20
train_losses, test_losses = [], []

for epoch in range(1, EPOCHS + 1):
    model.train()
    total_train_loss = 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        preds = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        optimizer.step()
        total_train_loss += loss.item() * xb.size(0)

    avg_train_loss = total_train_loss / len(train_loader.dataset)
    train_losses.append(avg_train_loss)

    # Evaluate on test data
    model.eval()
    total_test_loss = 0
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            preds = model(xb)
            loss = criterion(preds, yb)
            total_test_loss += loss.item() * xb.size(0)

    avg_test_loss = total_test_loss / len(test_loader.dataset)
    test_losses.append(avg_test_loss)

    print(f"Epoch {epoch:02d} | Train MSE: {avg_train_loss:.6f} | Test MSE: {avg_test_loss:.6f}")
```


### Plot training and test loss

Visualizing the loss curves helps diagnose overfitting or underfitting. Ideally, both training and test losses should decrease and converge.


```

plt.figure(figsize=(8, 4))
plt.plot(range(1, EPOCHS + 1), train_losses, 'o-', label='Training Loss')
plt.plot(range(1, EPOCHS + 1), test_losses, 'o-', label='Test Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.title('Training and Test Loss')
plt.legend()
plt.grid(True)
plt.show()
```


## 7. Forecast a sample sequence

After training, pick a sample from the test set to see how the model predicts the next seven days. The predictions are de‑normalized using the stored mean and standard deviation so that the results are expressed in physical units.


```

# Sample index to visualize
sample_idx = 0  # Change to view other samples

# Prepare the sample
X_sample = X_test_tensor[sample_idx].unsqueeze(0).to(device)
y_true   = y_test_tensor[sample_idx].numpy()

# Make prediction
model.eval()
with torch.no_grad():
    y_pred_norm = model(X_sample).cpu().numpy()[0]

# De-normalize
y_pred = y_pred_norm * std + mean

# True values in original units
y_true_orig = y_true * std + mean

# Plot
plt.figure(figsize=(12, 4))
for i, name in enumerate(['AWND', 'PRCP', 'TAVG']):
    plt.subplot(1, 3, i+1)
    plt.plot(y_true_orig[:, i], 'g-o', label='True')
    plt.plot(y_pred[:, i], 'b--o', label='Predicted')
    plt.title(name)
    plt.xlabel('Day ahead')
    plt.legend()

plt.tight_layout()
plt.show()
```


## 8. Saving your model

To submit your model for evaluation, save the class definition to a Python file (e.g., `WeatherRNN.py`) and the trained weights to a file (e.g., `WeatherRNN.pt`). Below is an example of how to do this. Make sure to execute these cells after training is complete.


```

# Example: save the model architecture to a Python file
model_code = '''
import torch
from torch import nn

class WeatherRNN(nn.Module):
    def __init__(self, n_future_days, input_size=3, hidden_size=64, num_layers=1):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )
        self.fc = nn.Linear(hidden_size, n_future_days * input_size)
        self.n_future_days = n_future_days
        self.input_size = input_size

    def forward(self, x):
        out, _ = self.lstm(x)
        last_hidden = out[:, -1, :]
        out_fc = self.fc(last_hidden)
        return out_fc.view(-1, self.n_future_days, self.input_size)
'''

# Write the model class to WeatherRNN.py
with open('WeatherRNN.py', 'w') as f:
    f.write(model_code)

print('Model class saved to WeatherRNN.py')

# Save the trained weights
weights_path = 'WeatherRNN.pt'
torch.save(model.state_dict(), weights_path)
print(f'Model weights saved to {weights_path}')
```


## 9. Example instructions for using the trained model

When scoring the homework, the instructor will provide a 30‑day window of test data. To generate a forecast:

1. **Normalize** the test data using the same means and standard deviations (`mean` and `std`) computed from the training set.
2. **Load** the model architecture and weights.
3. **Make a prediction** on the normalized data.
4. **De‑normalize** the prediction to obtain values in physical units.

Below is an example code snippet showing how to perform these steps. You should adjust the file names to match your model files.


```

# Load 30-day test data from a binary file provided at evaluation time
# (In practice, replace 'Test_Weather_Data' with the actual filename.)
test_data = np.float32(np.fromfile('Test_Weather_Data', '>f4').reshape((1, 30, 3)))

# Normalize using training means and stds
test_data_norm = test_data.copy()
test_data_norm[:, :, 0] = (test_data_norm[:, :, 0] - mean[0]) / std[0]
test_data_norm[:, :, 1] = (test_data_norm[:, :, 1] - mean[1]) / std[1]
test_data_norm[:, :, 2] = (test_data_norm[:, :, 2] - mean[2]) / std[2]

# Convert to torch tensor
input_tensor = torch.tensor(test_data_norm).float().to(device)

# Load the saved model architecture and weights
from WeatherRNN import WeatherRNN
loaded_model = WeatherRNN(n_future_days=7, input_size=3, hidden_size=64, num_layers=1).to(device)
loaded_model.load_state_dict(torch.load('WeatherRNN.pt', map_location=device))
loaded_model.eval()

# Make forecast
with torch.no_grad():
    forecast_norm = loaded_model(input_tensor).cpu().numpy()

# De-normalize
forecast = forecast_norm.copy()
forecast[:, :, 0] = forecast[:, :, 0] * std[0] + mean[0]
forecast[:, :, 1] = forecast[:, :, 1] * std[1] + mean[1]
forecast[:, :, 2] = forecast[:, :, 2] * std[2] + mean[2]

# forecast now contains the predicted AWND, PRCP, and TAVG for the next 7 days in physical units
print('Predicted 7-day forecast:
', forecast)
```
