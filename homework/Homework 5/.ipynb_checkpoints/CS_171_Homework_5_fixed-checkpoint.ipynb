#### Precautions
- This is a real dataset from the wild. This means that there will be missing data and bad values. Be sure to remove days without data or with bad values. Plotting the data will help to identify these periods.

```python

# Compute climatology and per-feature anomalies
# Keep only the AWND, PRCP, and TAVG columns and drop missing values
data = df[['AWND', 'PRCP', 'TAVG']].dropna().reset_index(drop=True).values

num_days = len(data)
num_years = num_days // 365  # number of full years in the record

# Build a 365-day climatology (mean annual cycle) for each variable
climatology = np.zeros((365, 3))
for year in range(num_years):
    year_slice = data[year*365:(year+1)*365]
    climatology += year_slice
climatology /= num_years

# Construct a climatology time series and anomaly time series
data_clim = np.zeros_like(data)
data_anom = np.zeros_like(data)
for year in range(num_years):
    start = year * 365
    end = (year + 1) * 365
    data_clim[start:end] = climatology
    data_anom[start:end] = data[start:end] - climatology

# Handle leftover days if the record isn’t a multiple of 365
days_left = num_days % 365
if days_left != 0:
    leftover_start = num_years * 365
    data_clim[leftover_start:] = climatology[:days_left]
    data_anom[leftover_start:] = data[leftover_start:] - climatology[:days_left]

# Compute per-feature mean and standard deviation of anomalies
mean_anomaly = data_anom.mean(axis=0)
std_anomaly = data_anom.std(axis=0)

# Normalize anomalies (per feature)
data_anom_scaled = (data_anom - mean_anomaly) / std_anomaly

# Save these normalization parameters for later use
a_mean_wind, a_mean_precip, a_mean_temp = mean_anomaly
a_std_wind, a_std_precip, a_std_temp = std_anomaly

# You can now build sequences from data_anom_scaled for training
```

## Instructions for Running Your Model
To test your model, Mike is going to need to know how to use it! Here is an example dataset, model and set of instructions to see how this section should operate. You should update this section depending on the creation of your model:

```python
# read in the test data
# this data will be replaced with real data
# one week after the assignment due data
test_data = np.float32(np.fromfile('Test_Weather_Data','>f4').reshape((1,30,3)))

# read in the model
from WeatherRNN import WeatherRNN
model = WeatherRNN(7,3).to(device)
state_dict = torch.load('WeatherRNN', map_location=torch.device(device));
model.load_state_dict(state_dict)
model.eval();
```

#### Example Instructions
To use my model, first normalize the data with the following parameters:

```python

# Example normalization parameters (per-feature anomalies)
# These values are computed from the anomaly time series in the training data
mean_wind_anom   = a_mean_wind
std_wind_anom    = a_std_wind
mean_precip_anom = a_mean_precip
std_precip_anom  = a_std_precip
mean_temp_anom   = a_mean_temp
std_temp_anom    = a_std_temp

# Use these to normalize the anomalies for each variable in the test data
```

```python

# Normalize test data using the climatology and anomaly parameters
# test_data is shape (1, 30, 3)

# Copy and prepare to subtract climatology for each day in the 30-day window
norm_data = test_data.copy()

# You must specify the starting day-of-year index for your 30-day window.
# If your training/test split ends at a particular day-of-year, adjust accordingly.
start_day_index = 0  # modify based on your data split if needed

# Subtract climatology (annual cycle) to get anomalies
for i_day in range(30):
    doy = (start_day_index + i_day) % 365
    norm_data[:, i_day, 0] -= climatology[doy][0]  # AWND
    norm_data[:, i_day, 1] -= climatology[doy][1]  # PRCP
    norm_data[:, i_day, 2] -= climatology[doy][2]  # TAVG

# Apply anomaly normalization (per feature)
norm_data[:, :, 0] = (norm_data[:, :, 0] - mean_wind_anom)   / std_wind_anom
norm_data[:, :, 1] = (norm_data[:, :, 1] - mean_precip_anom) / std_precip_anom
norm_data[:, :, 2] = (norm_data[:, :, 2] - mean_temp_anom)   / std_temp_anom

# Convert to tensor
test_data_tensor = torch.tensor(norm_data).float().to(device)
```

Then, pass the data to the model to make a forecast:

```python
forecast = model(test_data_tensor).to(device)
```

```python
forecast = forecast.cpu().detach().numpy()
```

To make a forecast of absolute values, de-normalize the data:

```python

# De-normalize the forecast
# forecast is shape (1, 7, 3) with scaled anomalies

# Add back the per-feature means and multiply by std
forecast[:, :, 0] = forecast[:, :, 0] * std_wind_anom  + mean_wind_anom
forecast[:, :, 1] = forecast[:, :, 1] * std_precip_anom + mean_precip_anom
forecast[:, :, 2] = forecast[:, :, 2] * std_temp_anom   + mean_temp_anom

# Add back climatology for the forecast days
for i_day in range(7):
    doy = (start_day_index + 30 + i_day) % 365
    forecast[:, i_day, 0] += climatology[doy][0]
    forecast[:, i_day, 1] += climatology[doy][1]
    forecast[:, i_day, 2] += climatology[doy][2]
```

Then, plot everything to visualize the forecast:

```python
fig = plt.figure(figsize=(8,6))

plt.subplot(3,1,1)
plt.plot(test_data[0,:,0], color='purple')
plt.plot(np.arange(30,37), forecast[0,:,0], '--', color='purple')
plt.ylabel('Wind Speed (kts.)')

plt.subplot(3,1,2)
plt.plot(test_data[0,:,1], color='blue')
plt.plot(np.arange(30,37), forecast[0,:,1], '--', color='blue')
plt.ylabel('Precip (in.)')

plt.subplot(3,1,3)
plt.plot(test_data[0,:,2], color='red')
plt.plot(np.arange(30,37), forecast[0,:,2], '--', color='red')
plt.ylabel('Temperature ($^{\\circ}$F)')
plt.xlabel('Days')
plt.show()
```
