```python
# 10. 
class NN(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(300, 200)
        self.fc2 = nn.Linear(200, 1000)
        self.fc3 = nn.Linear(1000, 50)

    def forward(self, x):
        x = torch.sigmoid(self.fc1(x))                      # activation function
        x = torch.sigmoid(self.fc2(x))                      # activation function
        x = torch.sigmoid(self.fc3(x))                      # activation function
        return x

# --------------------------------------------------------------------------

model = NN().to(device)

ds = TensorDataset(X_train_tensor, y_train)
loader = DataLoader(ds, batch_size=64, shuffle=True)

criterion = nn.MSELoss()                                    # gradient descent definition
optimizer = torch.optim.SGD(model.parameters(), lr=0.5)     # gradient descent definition

model.train()

for epoch in range(10):
    for x, y in loader:
        x, y = x.to(device), y.to(device)

        p = model(x)                                        # forward method application

        loss = criterion(p, y)
        optimizer.zero_grad()
        loss.backward()                                     # backprop application
        optimizer.step()                                    # gradient descent application
```
