There's a bunch of magic happening in the middle that nobody really knows the exact details of but just like how we learn, neural networks learn what inputs correspond to what outputs and tunes itself to achieve the best result it can. 

## Problem 11.2 - Model Purposes

For each of the NNs listed above, explain what type of data they are designed to model.

Feed-forward NN: Lists (vectors) of numbers.

Convolutional NN: Data that is arranged in 2 dimensional grids, like images.

Recurrent NN: Data that is time-based or otherwise sequential.

## Problem 11.3 - Model Types

Your friend is starting to understand a little bit better. Now, in 2-3 sentences, explain how a CNN differs from an RNN. How do each of these differ from a "regular" or "feed-forward" neural network?

A CNN aims to learn spatially from the data, trying to figure out how features influence each other in a 2D form. Meanwhile an RNN aims to learn from how previous states compare to current states. A regular NN just tries to figure out the best combination of weights in a list and are far less complex.
