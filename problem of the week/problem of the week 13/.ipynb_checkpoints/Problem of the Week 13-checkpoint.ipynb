## Problem 13.2 - k-Nearest Neighbors vs k-Mean Clustering

k-Nearest Neighbors is a supervised algorithm, while k-Means Clustering is unsupervised. the k in kNN is for the number of nearby points, while the k in k-Mean clustering refers to the number of clusters. kNN is meant to predict future labels for labeled data while k-Means clustering is supposed to just cluster existing data.
